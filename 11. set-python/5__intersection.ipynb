intersection set

```python
# English 

# Intersection finds the elements that are common in both sets.

# Hindi 

# Intersection ka matlab hai 2 Sets mein jo values common hain, sirf unhi values ko nikalna.
```

Simple Example

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

result = A.intersection(B)

print(result)

# jo same same rhega wahi output dega.
```

```python
#    UNION        → dono sets ki saari unique values
#    INTERSECTION → dono sets ki sirf common values
```

5 Example

```python
# 1. Simple numbers
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.intersection(B))
# Output: {3, 4}


# 2. Students ke common subjects
A = {"Python", "HTML", "CSS"}
B = {"HTML", "CSS", "JavaScript"}

print(A.intersection(B))
# Output: {'HTML', 'CSS'}


# 3. Common numbers
A = {10, 20, 30, 40}
B = {20, 40, 60, 80}

print(A.intersection(B))
# Output: {20, 40}


# 4. Teen sets ka intersection
A = {1, 2, 3, 4, 5}
B = {2, 3, 4, 6}
C = {3, 4, 7}

print(A.intersection(B, C))
# Output: {3, 4}


# 5. Koi common value nahi
A = {1, 2, 3}
B = {4, 5, 6}

print(A.intersection(B))
# Output: set()
```
