remove set

```python
# English

# remove() is used to remove a specific value from a Set.

# Hindi

# remove() ka use Set ke andar se kisi specific value ko delete karne ke liye hota hai.
```

syntax

```python
# set_name.remove(value)
```

5 Example 

```python
# 1. Number remove karna
a = {10, 20, 30, 40}
a.remove(30)
print(a)
# {10, 20, 40}


# 2. String remove karna
a = {"Python", "HTML", "CSS"}
a.remove("HTML")
print(a)
# {"Python", "CSS"}


# 3. Last value remove karna
a = {10, 20, 30}
a.remove(30)
print(a)
# {10, 20}


# 4. Value present nahi hai
a = {10, 20, 30}
a.remove(50)
print(a)
# KeyError
```

```python
#    remove() → Set se specific value delete karta hai.
#    Value Set mein nahi hai → Error (KeyError) aayega. 
#    Set mein index nahi hota, isliye remove(2) ka matlab index 2 remove karna nahi, balki value 2 remove karna hai.
```
