set union

```python
# English

# Union means combining the elements of two sets into one set, while removing duplicate values.

# Hindi

# Union ka matlab hai 2 sets ki saari unique values ko ek saath combine karna.
# Agar same value dono sets mein hai, to wo sirf ek baar aayegi.
```

syntax

```python
# set1.union(set2)
```

Example

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

C = A.union(B)

print(C)
```

5 Example

```python
# 1. Simple Union
A = {1, 2, 3}
B = {3, 4, 5}

print(A.union(B))
# Output: {1, 2, 3, 4, 5}


# 2. Students ke subjects
A = {"Python", "HTML"}
B = {"CSS", "Python"}

print(A.union(B))
# Output: {'Python', 'HTML', 'CSS'}


# 3. Numbers
A = {10, 20, 30}
B = {30, 40, 50}

print(A.union(B))
# Output: {10, 20, 30, 40, 50}


# 4. Multiple sets
A = {1, 2}
B = {3, 4}
C = {5, 6}

print(A.union(B, C))
# Output: {1, 2, 3, 4, 5, 6}


# 5. Duplicate values
A = {10, 20, 30}
B = {20, 30, 40}

C = A.union(B)

print(C)
# Output: {10, 20, 30, 40}
```
