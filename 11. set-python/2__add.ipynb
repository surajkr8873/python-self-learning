add set

```python
# English

# add() is used to add a new value to a Set.

# Hindi

# add() ka use Set ke andar ek nayi value add karne ke liye hota hai.
```

```python
# set_name.add(value)
```

5 Examples

```python
# 1. Number add karna
a = {10, 20, 30}
a.add(40)
print(a)
# {10, 20, 30, 40}


# 2. String add karna
a = {"Python", "HTML"}
a.add("CSS")
print(a)
# {'Python', 'HTML', 'CSS'}


# 3. Duplicate value add karna
a = {10, 20, 30}
a.add(20)
print(a)
# {10, 20, 30}
# Duplicate value add nahi hogi


# 4. Empty set mein value add karna
a = set()
a.add(100)
print(a)
# {100}


# 5. Multiple values add karna
a = {1, 2}
a.add(3)
a.add(4)
a.add(5)
print(a)
# {1, 2, 3, 4, 5}
```

```python
#    add() → Set mein 1 value add karta hai ✅
#    Duplicate value → add nahi hoti ❌
#    add() ek time par sirf ek value leta hai.
#    Set unordered hai, isliye output ka order fixed nahi hota.
```
