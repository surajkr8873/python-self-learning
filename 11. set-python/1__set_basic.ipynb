What is set datatype in python ?

```python
# English

# A set is an unordered collection of unique values.

# Hindi

# Set ek collection hai jisme unique (alag-alag) values store hoti hain. Set mein duplicate values nahi hoti.
```

Set kaise banate hain?

```python
# Set ko normally curly brackets { } ke andar banate hain.

numbers = {10, 20, 30, 40}

print(numbers)
```

Duplicate Values >> remove

```python
numbers = {10, 20, 10, 30, 20}

print(numbers)
```

Set ki Basic Properties

```python
# 1__Unordered → fixed index/order par depend nahi karte.
# 2__Unique → duplicate values allowed nahi.
# 3__Mutable → Set mein values add/remove kar sakte hain.
# 4__Indexing nahi hoti → set[0] 
# 5__Different data types store kar sakta hai.
```

Example

```python
data = {10, "Suraj", 18.5, True}

print(data)
```

Empty Set

```python
# a = {}

# Ye empty dictionary hai, empty set nahi.

# Empty set banane ke liye:

a = set()

print(a)
```

```python
# Set = { } → unique values

a = {10, 20, 30, 20}

print(a)
# {10, 20, 30}
```

5 Example 

```python
# 1. Simple Set
a = {10, 20, 30, 40}
print(a)
# {10, 20, 30, 40}


# 2. Duplicate values automatically remove
a = {10, 20, 10, 30, 20}
print(a)
# {10, 20, 30}


# 3. Different data types
a = {10, "Suraj", 18.5, True}
print(a)


# 4. Set mein value add karna
a = {10, 20, 30}
a.add(40)
print(a)
# {10, 20, 30, 40}


# 5. Set se value remove karna
a = {10, 20, 30, 40}
a.remove(30)
print(a)
# {10, 20, 40}
```
