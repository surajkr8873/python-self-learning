set difference

```python
# English 

# Difference finds the elements that are present in the first set but not in the second set.

# Hindi 

# Difference ka matlab hai: pehle Set mein jo values hain, lekin doosre Set mein nahi hain, unko nikalna.
```

Example

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.difference(B))

# jo phle wale set me ho dusre me na ho.

print(B.difference(A))

# jo dusre  wale set me ho phlle me na ho
```

Shortcut Operator

Difference ko - se bhi likh sakte ho:

```python
A = {1, 2, 3}
B = {2, 3, 4}

print(A - B)
# {1}
```

5 Example

```python
# 1. Simple numbers
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.difference(B))
# Output: {1, 2}


# 2. Strings
A = {"Python", "HTML", "CSS"}
B = {"HTML", "JavaScript"}

print(A.difference(B))
# Output: {'Python', 'CSS'}


# 3. Common numbers ko hataana
A = {10, 20, 30, 40}
B = {20, 40, 50}

print(A.difference(B))
# Output: {10, 30}


# 4. Difference ka direction
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A.difference(B))
# Output: {1, 2}

print(B.difference(A))
# Output: {5, 6}


# 5. Koi common value nahi
A = {10, 20, 30}
B = {40, 50, 60}

print(A.difference(B))
# Output: {10, 20, 30}
```
