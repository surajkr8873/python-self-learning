remove disctionary

```python
# Dictionary se data remove karne ke mainly 3 important methods hain:

# pop() → specific key remove
# popitem() → last key-value pair remove
# del → specific key remove
```

```python
# pop()

# pop() se specific key aur uski value remove hoti hai.

# Syntax
```

```python
# dictionary.pop("key")
```

Example

```python
student = {
    "name": "Suraj",
    "age": 18,
    "course": "Python"
}

student.pop("age")

print(student)
```

popitem()>>

popitem() dictionary ka last key-value pair remove karta hai.

```python
student = {
    "name": "Suraj",
    "age": 18,
    "course": "Python"
}

student.popitem()

print(student)
```

del

del se bhi specific key-value pair delete kar sakte hain.

```python
student = {
    "name": "Suraj",
    "age": 18,
    "course": "Python"
}

del student["age"]

print(student)
```

5 Example

```python
# 1. pop()

student = {"name": "Suraj", "age": 18}
student.pop("age")
print(student)

# {'name': 'Suraj'}


# 2. pop() se course remove

student = {"name": "Suraj", "course": "Python"}
student.pop("course")
print(student)

# {'name': 'Suraj'}


# 3. popitem()

student = {"name": "Suraj", "age": 18, "city": "Munger"}
student.popitem()
print(student)

# {'name': 'Suraj', 'age': 18}


# 4. del

student = {"name": "Suraj", "age": 18}
del student["age"]
print(student)

# {'name': 'Suraj'}


# 5. Puri dictionary delete

student = {"name": "Suraj", "age": 18}
del student

# Dictionary completely deleted
```

```python
#                      Method	                             Kaam

#                    pop("key")	                       Specific key remove
#                    popitem()                         Last pair remove
#                    del dict["key"]	               Specific key remove
#                    del dict	                       Puri dictionary delete
```
