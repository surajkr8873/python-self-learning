5 Example

```python
# 1. New age add karna
student = {"name": "Suraj"}

student["age"] = 18

print(student)
# {'name': 'Suraj', 'age': 18}


# 2. Course add karna
student = {"name": "Suraj", "age": 18}

student["course"] = "Python"

print(student)
# {'name': 'Suraj', 'age': 18, 'course': 'Python'}


# 3. Email add karna
student = {"name": "Suraj"}

student["email"] = "suraj@gmail.com"

print(student)
# {'name': 'Suraj', 'email': 'suraj@gmail.com'}


# 4. Number add karna
numbers = {"a": 10, "b": 20}

numbers["c"] = 30

print(numbers)
# {'a': 10, 'b': 20, 'c': 30}


# 5. Multiple data add karna
student = {"name": "Suraj"}

student["age"] = 18
student["course"] = "Python"
student["city"] = "Munger"

print(student)
# {'name': 'Suraj', 'age': 18, 'course': 'Python', 'city': 'Munger'}
```
