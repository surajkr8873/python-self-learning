disctionary update

```python
# English

# update() is used to add new key-value pairs or change existing values in a dictionary.

# Hindi

# update() ka use dictionary mein naya data add karne ya existing data ki value change karne ke liye hota hai.
```

Syntax

```python
# dictionary.update({
#     "key": value
# })
```

Example

```python
student = {
    "name": "Suraj",
    "age": 18
}

student.update({
    "age": 19,
    "course": "Python"
})

print(student)
```

5 Example

```python
# 1. Existing value update
student = {"name": "Suraj", "age": 18}

student.update({"age": 19})

print(student)
# {'name': 'Suraj', 'age': 19}


# 2. New key-value add
student = {"name": "Suraj"}

student.update({"age": 18})

print(student)
# {'name': 'Suraj', 'age': 18}


# 3. Multiple values update
student = {"name": "Suraj", "age": 18}

student.update({
    "age": 19,
    "course": "Python"
})

print(student)
# {'name': 'Suraj', 'age': 19, 'course': 'Python'}


# 4. Multiple dictionaries ka data combine
a = {"name": "Suraj"}
b = {"age": 18, "course": "Python"}

a.update(b)

print(a)
# {'name': 'Suraj', 'age': 18, 'course': 'Python'}


# 5. Existing value replace
student = {
    "name": "Suraj",
    "city": "Munger"
}

student.update({"city": "Patna"})

print(student)
# {'name': 'Suraj', 'city': 'Patna'}
```

update() → Add bhi kar sakta hai
          Update bhi kar sakta hai
