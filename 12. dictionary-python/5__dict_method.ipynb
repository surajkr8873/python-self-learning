```python
# Dictionary mein kuch built-in methods hote hain jo data ko access, check, update aur manage karne mein help karte hain.
```

```python
#                       Method                                    	Kaam

#                      keys()	                              Saari keys deta hai
#                     values()	                              Saari values deta hai
#                     items()	                              Key + Value pairs deta hai
#                     get()	                                  Kisi key ki value deta hai
#                     clear()	                              Puri dictionary empty karta hai
#                     copy()	                              Dictionary ki copy banata hai
```

5 Example

```python
student = {
    "name": "Suraj",
    "age": 18,
    "course": "Python"
}

print(student.keys())
# Saari keys

print(student.values())
# Saari values

print(student.items())
# Key-value pairs

print(student.get("name"))
# Suraj

student.clear()
print(student)
# {}
```
