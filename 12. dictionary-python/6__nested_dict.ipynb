nested disctionary

```python
# English 

# A Nested Dictionary means a dictionary inside another dictionary.

# Hindi 

# Jab ek dictionary ke andar ek aur dictionary hoti hai, use Nested Dictionary kehte hain.
```

Basic Example

```python
students = {
    "student1": {
        "name": "Suraj",
        "age": 18
    },

    "student2": {
        "name": "Rahul",
        "age": 19
    }
}

print(students)
```

```python
# phle student1 ka data dega uske baad student2 ka
```

5 Example

```python
# 1. Student
student = {
    "info": {
        "name": "Suraj",
        "age": 18
    }
}

print(student["info"]["name"])
# Suraj


# 2. Multiple students
students = {
    "student1": {"name": "Suraj", "age": 18},
    "student2": {"name": "Rahul", "age": 19}
}

print(students["student2"]["name"])
# Rahul


# 3. Course information
course = {
    "python": {
        "duration": "3 months",
        "level": "Beginner"
    }
}

print(course["python"]["level"])
# Beginner


# 4. Employee
employee = {
    "details": {
        "name": "Aman",
        "salary": 25000
    }
}

print(employee["details"]["salary"])
# 25000


# 5. Nested dictionary mein update
student = {
    "info": {
        "name": "Suraj",
        "age": 18
    }
}

student["info"]["age"] = 19


# {'info': {'name': 'Suraj', 'age': 19}}
```
