What is disctionary datatype in python

```python
# English

# A Dictionary is a collection of data stored in key-value pairs.

# Hindi

# Dictionary mein data Key : Value ke form mein store hota hai.
```

Example

```python
student = {"name": "Suraj","age": 18,"course": "Python"}

print(student)


#    "name"   → Key
#    "Suraj"  → Value

#    "age"    → Key
#     18       → Value

#    "course" → Key
#    "Python" → Value
```

```python
#    1__Dictionary → { } braces use karti hai
#    2__Data → Key : Value pair mein hota hai
#    3__Key se value ko access karte hain
#    4__Keys unique honi chahiye
#    5__Values duplicate ho sakti hain
```
