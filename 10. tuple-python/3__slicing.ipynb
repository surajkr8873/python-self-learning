Slicing in tuple

```python
# English

# Tuple slicing is used to get a part of a tuple.

# Hindi

# Tuple me se kuch values ko ek saath range/index ke through nikalna slicing kehlata hai.
```

Syntax

```python
# tuple[start : stop]

#    start → jahan se start karna hai
#    stop → jahan tak jana hai
#    stop wala index include nahi hota
```

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[1:4])
```

Step ke saath slicing

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[0:6:2])
```

5 Examples

```python
# 1
a = (10, 20, 30, 40, 50)
print(a[0:3])
# (10, 20, 30)


# 2
a = (10, 20, 30, 40, 50)
print(a[2:5])
# (30, 40, 50)


# 3
a = (10, 20, 30, 40, 50)
print(a[:3])
# (10, 20, 30)


# 4
a = (10, 20, 30, 40, 50)
print(a[2:])
# (30, 40, 50)


# 5
a = (10, 20, 30, 40, 50, 60)
print(a[0:6:2])
# (10, 30, 50)
```
