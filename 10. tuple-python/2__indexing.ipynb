indexing in tuple

```python
# English

# Indexing is used to access a specific element from a tuple using its index number.

# Hindi

# Tuple me kisi particular value ko index number ke through nikalna indexing kehlata hai.
```

Example

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[0])
```

Negative Indexing

```python
numbers = (10, 20, 30, 40, 50)

print(numbers[-1])
```

5 Examples

```python
# 1
a = (10, 20, 30, 40)
print(a[0])
# 10


# 2
a = (10, 20, 30, 40)
print(a[2])
# 30


# 3
a = ("Python", "HTML", "CSS")
print(a[1])
# HTML


# 4
a = (10, 20, 30, 40)
print(a[-1])
# 40


# 5
a = (10, 20, 30, 40)
print(a[-3])
# 20
```
