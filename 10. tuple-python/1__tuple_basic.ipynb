tuple

What is tuple datatype in python ?

```python
# English

# A tuple is an ordered collection of multiple values that cannot be changed after it is created.

# Hindi

# Tuple ek collection hai jisme hum multiple values store kar sakte hain, lekin tuple banne ke baad uski values ko change nahi kar sakte.
```

Tuple kaise banate hain?

```python
numbers = (10, 20, 30, 40)

print(numbers)
```

Tuple ki Basic Properties

```python
#    1__Ordered → values ka order maintain rehta hai.
#    2__Immutable → banne ke baad values change nahi kar sakte.
#    3__Duplicates allowed → same value multiple times ho sakti hai.
#    4__Indexing allowed → index 0 se start hota hai.
#    5__Different data types store kar sakte hain.
```

Example

```python
data = (10, "Suraj", 18.5, True)

print(data)
```

```python
# List → [ ] → change kar sakte hain 

# Tuple → ( ) → change nahi kar sakte
```
