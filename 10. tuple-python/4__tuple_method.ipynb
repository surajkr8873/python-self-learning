tuple method in python

```python
# English

# count() tells how many times a value appears in a tuple.

# Hindi

# count() batata hai ki koi value tuple mein kitni baar present hai.
```

```python
a = (10, 20, 10, 30, 10)

print(a.count(10))
```

indexx()

```python
# 🇬🇧English

# index() returns the index position of a value.

# Hindi

# index() kisi value ka index/position batata hai.


a = (10, 20, 30, 40)

print(a.index(30))
```

Ek-ek Example

```python
# 1. count()
a = (10, 20, 10, 30, 10)

print(a.count(10))
# Output: 3


# 2. index()
a = (10, 20, 30, 40)

print(a.index(30))
# Output: 2
```

```python
# count() → value kitni baar hai
# index() → value ka index kya hai
```

```python
# 1. append() → end mein value add
a = [10, 20, 30]
a.append(40)

print(a)
# [10, 20, 30, 40]


# 2. insert() → kisi index par value add
a = [10, 20, 40]
a.insert(2, 30)

print(a)
# [10, 20, 30, 40]


# 3. remove() → particular value remove
a = [10, 20, 30]
a.remove(20)

print(a)
# [10, 30]


# 4. sort() → ascending order mein arrange
a = [30, 10, 20]
a.sort()

print(a)
# [10, 20, 30]
```
