```python
num = [10, 5, 25, 8, 20]

largest = num[0]
second = num[0]

for i in num:
    if i > largest:
        second = largest
        largest = i
    elif i > second and i != largest:
        second = i

print(second)
```

```python
num = [0, 1, 0, 3, 12, 0, 5]
answer = []

for i in num:
    if i != 0:
        answer.append(i)

for i in num:
    if i == 0:
        answer.append(i)

print(answer)
```

```python
def move_zero(arr):
    answer = []

    for num in arr:
        if num != 0:
            answer.append(num)

    for num in arr:
        if num == 0:
            answer.append(num)
    return answer

print(move_zero([0 , 2 , 5 , 0 , 5 , 8 , 0]))
```
