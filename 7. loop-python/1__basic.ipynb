What is loops in python ?

```python
English

# A loop is used to repeat a block of code multiple times.

# Instead of writing the same code again and again, we use a loop.

# Hindi

# Loop ka use kisi code ko baar-baar repeat karne ke liye kiya jata hai.

# Agar hume same kaam 10, 100 ya 1000 baar karna ho, to baar-baar code likhne ki zarurat nahi hoti. Hum loop ka use karte hain.
```

simple Example>> without loop

```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")

# Ye code unnecessary long ho gaya.
```

With Loop

```python
for i in range(5):
    print("hello")


# loops ka use krke hum ik hi code me multiple chije print kar skte hai hume baar baar ik hi code ko likhne ka jarurat nahi pdta hai. 
```

loops kaa basic ideas

```python
#                     Start
#                      ↓
#              Condition / Range
#                      ↓
#                  Code execute
#                      ↓
#                   Repeat
#                      ↓
#               Condition false
#                      ↓
#                     Stop

# loops tab tak chlta rhega  jab tak loops ke andar kaa condition False  nahi ho jata hai. 
```

Python mein mainly 2 types ke loops hote hain:

```python
# 1. for loop
# 2. while loop
```

Jab hume pata ho ki kitni baar repeat karna hai, tab for loop commonly use hota hai.

```python
for i in range(5):
    print("Hello")

# isme loop 5 baar chla 
```

While loop

```python
# Jab loop ko condition ke basis par chalana ho, tab while loop use karte hain.

i = 1

while i <= 5:
    print("Hello")
    i += 1
```

1.iteration

```python
# Loop ka ek baar chalna = one iteration.

for i in range(3):
    print(i)

# yaha loop 3 baar chalega matlab 3 iiteration. 
```

repeatation.

```python
# Same code ko baar-baar chalana = repetition.

for i in range(5):
    print(i)

# Yahan i ko loop variable kehte hain.
```

5 Simple Examples

```python
# 1. Hello 3 times
for i in range(3):
    print("Hello")


# 2. 1 se 5 tak numbers
for i in range(1, 6):
    print(i)


# 3. 2 se 10 tak even numbers
for i in range(2, 11, 2):
    print(i)


# 4. 1 se 5 tak numbers ka square
for i in range(1, 6):
    print(i * i)


# 5. Same message 5 times
for i in range(5):
    print("Python is easy")
```
