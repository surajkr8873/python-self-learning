What is for loop in python ?

```python
# English

# A for loop is used to repeat a block of code for each item in a sequence or iterable.

# Hindi

# for loop ka use kisi list, string, tuple, set, dictionary, ya range() ke items ko ek-ek karke access karne ke liye hota hai.
```

basic syntax

```python
# for variable in iterable:
#     # code

for i in range(1, 6):
    print(i)

# Isme kya ho raha hai?

#    for → Python ko batata hai ki loop start karna hai.
#    i → loop variable hai.
#    in → items ko ek-ek karke lene ke liye.
#    range(1, 6) → 1 se 5 tak values dega.
#    : → loop ka block start.
#    print(i) → har iteration mein chalega.
```

For Loop kaise chalta hai?

```python
for i in range(1, 4):
    print(i)

# >>> i = 1 → print(1)
# >>> i = 2 → print(2)
# >>> i = 3 → print(3)

# i (itrable) > i jo hai wo ik ik karke sare value pr jayega or print krega
```

range() ke saath For Loop

range(stop)

```python
for i in range(5):
    print(i)
```

range(start, stop)

```python
for i in range(2, 6):
    print(i)
```

range(start, stop, step)

```python
for i in range(2, 11, 2):
    print(i)

# Yahan 2 ka matlab 2-2 se jump.
```

For Loop with List

```python
numbers = [10, 20, 30, 40]

for i in numbers:
    print(i)
```

For Loop with String

```python
name = "Suraj"

for char in name:
    print(char)
```

For Loop with Tuple

```python
numbers = (10, 20, 30)

for i in numbers:
    print(i)
```

For Loop with Set

```python
numbers = {10, 20, 30}

for i in numbers:
    print(i)
```

For Loop with Dictionary

```python
student = {"name": "Suraj", "age": 18}

for key in student:
    print(key)

student = {
    "name": "Suraj",
    "age": 18
}

print("out of first loop")

for key, value in student.items():
    print(key, value)
```

For Loop + Condition

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

For Loop + else

```python
for i in range(3):
    print(i)
else:
    print("Loop finished")

# else tab execute hota hai jab loop normally complete ho jata hai.
```

Nested For Loop

```python
for i in range(2):
    for j in range(3):
        print(i, j)
```

5 Example of for loop

```python
# 1. 1 to 10

for i in range(1,5 ):
    print(i)


# 2. Even numbers

for i in range(2, 15 , 2):
    print(i)


# 3. Odd numbers

for i in range(1, 20, 2):
    print(i)


# 4. List ke elements

numbers = [10, 20, 30, 40]

for i in numbers:
    print(i)


# 5. String ke characters

name = "Python"

for char in name:
    print(char)
```
