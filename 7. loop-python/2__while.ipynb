What is while loop in python ?

```python
# English

# A while loop repeatedly executes a block of code as long as a given condition is True.

# Hindi

# while loop kisi code ko tab tak baar-baar chalata hai jab tak condition True hoti hai.

# Jaise hi condition False hoti hai, loop stop ho jata hai.
```

Basic Syntax

```python
# while condition:
#     # code
```

Example

```python
i = 1

while i <= 5:
    print(i)
    i += 1

# Isme kya ho raha hai?

# Starting value 1.
# Check karo: kya i 5 se chhota ya equal hai?
# Current value print karo.
# i ko 1 se increase karo.
```

Reverse Counting

```python
i = 5

while i >= 1:
    print(i)
    i -= 1
```

Even Numbers with While

```python
num = 2

while num <= 10:
    print(num)
    num += 2
```

While + If

```python
i = 1

while i <= 10:
    if i % 2 == 0:
        print(i)
    i += 1


#    while → 1 se 10 tak loop
#    if → sirf even number check
#    % 2 == 0 → even number
```

While + User Input

```python
number = 1 # yaha hum input bhi le skte hai

while number <= 3:
    print("Hello")
    number += 1
```

5 Examples

```python
# 1. 1 to 5
i = 1

while i <= 5:
    print(i)
    i += 1


# 2. 5 to 1

i = 5

while i >= 1:
    print(i)
    i -= 1


# 3. Even numbers 1 to 20

i = 2

while i <= 20:
    print(i)
    i += 2


# 4. Odd numbers 1 to 10

i = 1

while i <= 10:
    print(i)
    i += 2


# 5. Table of 5

i = 1

while i <= 10:
    print(5 * i)
    i += 1
```
