Break in loop 

```python
# English

# break immediately stops the loop.

# Hindi

# Jab loop ke andar break milta hai, loop wahi par completely stop ho jata hai.
```

Example

```python
for i in range(1, 6):
    if i == 3:
        break
    print(i)

# Kyunki i == 3 hote hi break laga → loop stop.
```

5 Example of break in loop.

```python
# 1. for loop
for i in range(1, 6):
    if i == 3:
        break
    print(i)

# Output:
# 1
# 2


# 2. for loop
for i in range(1, 11):
    if i == 6:
        break
    print(i)

# Output:
# 1
# 2
# 3
# 4
# 5


# 3. while loop
i = 1

while i <= 10:
    if i == 5:
        break
    print(i)
    i += 1

# Output:
# 1
# 2
# 3
# 4


# 4. while loop
i = 10

while i >= 1:
    if i == 7:
        break
    print(i)
    i -= 1

# Output:
# 10
# 9
# 8


# 5. for loop
for i in range(1, 11):
    if i % 2 == 0:
        break
    print(i)

# Output:
# 1
```
