Continue in loop .

```python
# English

# continue skips the current iteration and moves to the next iteration.

# Hindi

# Jab loop ke andar continue aata hai, to us round ka baaki code skip ho jata hai aur loop next round mein chala jata hai.
```

5 Examples — for + while >> continue 

```python
# 1. for loop
for i in range(1, 6):
    if i == 3:
        continue
    print(i)

# Output:
# 1
# 2
# 4
# 5


# 2. for loop
for i in range(1, 6):
    if i == 2:
        continue
    print(i)

# Output:
# 1
# 3
# 4
# 5


# 3. while loop
i = 0

while i < 5:
    i += 1

    if i == 3:
        continue

    print(i)

# Output:
# 1
# 2
# 4
# 5


# 4. while loop
i = 0

while i < 5:
    i += 1

    if i == 4:
        continue

    print(i)

# Output:
# 1
# 2
# 3
# 5


# 5. for loop — even numbers skip
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)

# Output:
# 1
# 3
# 5
# 7
# 9
```
